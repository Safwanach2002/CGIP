```python
#11. Global and Intensity Thresholding

import cv2
import matplotlib.pyplot as plt

img = cv2.imread("/content/134324064010251857.jpg", 0)

ret, global_img = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)

intensity = cv2.threshold(img, 180, 255, cv2.THRESH_BINARY)[1]

plt.subplot(1, 2, 1)
plt.imshow(global_img, cmap='gray')
plt.title("Global Thresholding")

plt.subplot(1, 2, 2)
plt.imshow(intensity, cmap='gray')
plt.title("Intensity Thresholding")

plt.show()
```

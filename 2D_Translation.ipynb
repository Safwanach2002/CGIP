```python
#4. 2D Translation

import matplotlib.pyplot as plt

x = [1, 4, 4, 1, 1]
y = [1, 1, 3, 3, 1]

tx = 3
ty = 2

x1 = [i + tx for i in x]
y1 = [i + ty for i in y]

plt.plot(x, y, 'b-', label="Original")
plt.plot(x1, y1, 'r-', label="Translated")

plt.title("2D Translation")
plt.grid(True)
plt.legend()
plt.axis("equal")
plt.show()
```

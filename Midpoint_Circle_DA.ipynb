```python
#3. Midpoint Circle Drawing Algorithm

import matplotlib.pyplot as plt

r = 5
x = 0
y = r
p = 1 - r

X, Y = [], []

while x <= y:
    X.extend([x, -x, x, -x, y, -y, y, -y])
    Y.extend([y, y, -y, -y, x, x, -x, -x])

    x += 1

    if p < 0:
        p += 2 * x + 1
    else:
        y -= 1
        p += 2 * (x - y) + 1

plt.scatter(X, Y)
plt.title("Midpoint Circle Drawing")
plt.grid(True)
plt.axis("equal")
plt.show()
```

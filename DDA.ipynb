```python
#1. DDA Line Drawing Algorithm

import matplotlib.pyplot as plt

x1, y1 = 2, 3
x2, y2 = 10, 8

dx = x2 - x1
dy = y2 - y1

steps = max(abs(dx), abs(dy))

x_inc = dx / steps
y_inc = dy / steps

x, y = x1, y1
X, Y = [], []

for i in range(steps + 1):
    X.append(round(x))
    Y.append(round(y))
    x += x_inc
    y += y_inc

plt.plot(X, Y, 'k-')
plt.plot(X, Y, 'ko')
plt.title("DDA Line Drawing")
plt.grid(True)
plt.show()
```

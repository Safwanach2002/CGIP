```python
#6. 2D Scaling

import matplotlib.pyplot as plt

x = [1, 4, 4, 1, 1]
y = [1, 1, 3, 3, 1]

sx = 2
sy = 2

x1 = [i * sx for i in x]
y1 = [i * sy for i in y]

plt.plot(x, y, 'b-', label="Original")
plt.plot(x1, y1, 'r-', label="Scaled")

plt.title("2D Scaling")
plt.grid(True)
plt.legend()
plt.axis("equal")
plt.show()
```

```python
#1. DDA Line Drawing Algorithm

import matplotlib.pyplot as plt

x1, y1 = 2, 3
x2, y2 = 10, 8

dx = x2 - x1
dy = y2 - y1

steps = max(abs(dx), abs(dy))

x_inc = dx / steps
y_inc = dy / steps

x, y = x1, y1
X, Y = [], []

for i in range(steps + 1):
    X.append(round(x))
    Y.append(round(y))
    x += x_inc
    y += y_inc

plt.plot(X, Y, 'k-')
plt.plot(X, Y, 'ko')
plt.title("DDA Line Drawing")
plt.grid(True)
plt.show()
```

```python
#2. Bresenham's Line Drawing Algorithm

import matplotlib.pyplot as plt

x1, y1 = 2, 3
x2, y2 = 10, 8

dx = abs(x2 - x1)
dy = abs(y2 - y1)

sx = 1 if x1 < x2 else -1
sy = 1 if y1 < y2 else -1

p = dx - dy
x, y = x1, y1

X, Y = [], []

while True:
    X.append(x)
    Y.append(y)

    if x == x2 and y == y2:
        break

    e = 2 * p

    if e > -dy:
        p -= dy
        x += sx

    if e < dx:
        p += dx
        y += sy

plt.plot(X, Y, 'bo-')
plt.title("Bresenham Line Drawing")
plt.grid(True)
plt.axis("equal")
plt.show()
```

```python
#3. Midpoint Circle Drawing Algorithm

import matplotlib.pyplot as plt

r = 5
x = 0
y = r
p = 1 - r

X, Y = [], []

while x <= y:
    X.extend([x, -x, x, -x, y, -y, y, -y])
    Y.extend([y, y, -y, -y, x, x, -x, -x])

    x += 1

    if p < 0:
        p += 2 * x + 1
    else:
        y -= 1
        p += 2 * (x - y) + 1

plt.scatter(X, Y)
plt.title("Midpoint Circle Drawing")
plt.grid(True)
plt.axis("equal")
plt.show()
```

```python
#4. 2D Translation

import matplotlib.pyplot as plt

x = [1, 4, 4, 1, 1]
y = [1, 1, 3, 3, 1]

tx = 3
ty = 2

x1 = [i + tx for i in x]
y1 = [i + ty for i in y]

plt.plot(x, y, 'b-', label="Original")
plt.plot(x1, y1, 'r-', label="Translated")

plt.title("2D Translation")
plt.grid(True)
plt.legend()
plt.axis("equal")
plt.show()
```

```python
#5. 2D Rotation

import matplotlib.pyplot as plt
import math

x = [1, 4, 4, 1, 1]
y = [1, 1, 3, 3, 1]

angle = 90
a = math.radians(angle)

x1 = [i * math.cos(a) - j * math.sin(a) for i, j in zip(x, y)]
y1 = [i * math.sin(a) + j * math.cos(a) for i, j in zip(x, y)]

plt.plot(x, y, 'b-', label="Original")
plt.plot(x1, y1, 'r-', label="Rotated")

plt.title("2D Rotation")
plt.grid(True)
plt.legend()
plt.axis("equal")
plt.show()
```

```python
#6. 2D Scaling

import matplotlib.pyplot as plt

x = [1, 4, 4, 1, 1]
y = [1, 1, 3, 3, 1]

sx = 2
sy = 2

x1 = [i * sx for i in x]
y1 = [i * sy for i in y]

plt.plot(x, y, 'b-', label="Original")
plt.plot(x1, y1, 'r-', label="Scaled")

plt.title("2D Scaling")
plt.grid(True)
plt.legend()
plt.axis("equal")
plt.show()
```

```python
#7. Convert Image into Grayscale, Binary and RGB

import cv2
import matplotlib.pyplot as plt

img = cv2.imread("/content/134324064010251857.jpg")

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)[1]
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.subplot(1, 3, 1)
plt.imshow(gray, cmap='gray')
plt.title("Grayscale")

plt.subplot(1, 3, 2)
plt.imshow(binary, cmap='gray')
plt.title("Binary")

plt.subplot(1, 3, 3)
plt.imshow(rgb)
plt.title("RGB")

plt.show()
```

```python
#8. Histogram and Histogram Equalization

import cv2
import matplotlib.pyplot as plt

img = cv2.imread("/content/134324064010251857.jpg", 0)

equal = cv2.equalizeHist(img)

plt.subplot(2, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")

plt.subplot(2, 2, 2)
plt.imshow(equal, cmap='gray')
plt.title("Equalized Image")

plt.subplot(2, 2, 3)
plt.hist(img.ravel(), 256)
plt.title("Original Histogram")

plt.subplot(2, 2, 4)
plt.hist(equal.ravel(), 256)
plt.title("Equalized Histogram")

plt.show()
```

```python
#9. Sobel Edge Detection

import cv2
import matplotlib.pyplot as plt

img = cv2.imread("/content/134324064010251857.jpg", 0)

x = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)
y = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)

result = cv2.magnitude(x, y)

plt.imshow(result, cmap='gray')
plt.title("Sobel Edge Detection")
plt.show()
```

```python
#10. Prewitt Edge Detection

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("/content/134324064010251857.jpg", 0)

kx = np.array([[-1, 0, 1],
               [-1, 0, 1],
               [-1, 0, 1]])

ky = np.array([[-1, -1, -1],
               [0, 0, 0],
               [1, 1, 1]])

x = cv2.filter2D(img, -1, kx)
y = cv2.filter2D(img, -1, ky)

result = cv2.add(x, y)

plt.imshow(result, cmap='gray')
plt.title("Prewitt Edge Detection")
plt.show()
```

```python
#12. Image Translation

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("/content/134324064010251857.jpg")

rows, cols = img.shape[:2]

M = np.float32([[1, 0, 50],
                [0, 1, 30]])

result = cv2.warpAffine(img, M, (cols, rows))

plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
plt.title("Image Translation")
plt.show()
```

```python
#13. Image Rotation

import cv2
import matplotlib.pyplot as plt

img = cv2.imread("/content/134324064010251857.jpg")

rows, cols = img.shape[:2]

M = cv2.getRotationMatrix2D((cols/2, rows/2), 45, 1)

result = cv2.warpAffine(img, M, (cols, rows))

plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
plt.title("Image Rotation")
plt.show()
```

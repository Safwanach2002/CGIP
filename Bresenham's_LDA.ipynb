```python
#2. Bresenham's Line Drawing Algorithm

import matplotlib.pyplot as plt

x1, y1 = 2, 3
x2, y2 = 10, 8

dx = abs(x2 - x1)
dy = abs(y2 - y1)

sx = 1 if x1 < x2 else -1
sy = 1 if y1 < y2 else -1

p = dx - dy
x, y = x1, y1

X, Y = [], []

while True:
    X.append(x)
    Y.append(y)

    if x == x2 and y == y2:
        break

    e = 2 * p

    if e > -dy:
        p -= dy
        x += sx

    if e < dx:
        p += dx
        y += sy

plt.plot(X, Y, 'bo-')
plt.title("Bresenham Line Drawing")
plt.grid(True)
plt.axis("equal")
plt.show()
```
